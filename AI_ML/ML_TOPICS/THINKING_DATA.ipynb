```python
# FIRST IMPORTING THE JSON THEN ONLY U CAN DEAL WITH THE JSON DATA
import json
```

```python
# LOADING OF THE DATA --> in the form of function
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)

    return data
```

```python
# CALLING THE LOADING OF THE DATA FUNCTION AND PASSING THE NAME OF THE FILE THAT I WANT TO LOAD + READ HERE 
# USING A VARIABLE TO LOAD IT 
data = load_data("store_data.json")
print(data)
```

```python
# ACTIONS ON THE DATA
def clean_data(data):
    # MAKING A DICTIONARY KIND OF SYSTEM TO CLEAN THE RATING PART BY CONVERTING STRING --> TO NUMBERS
    # AS WE ALREADY KNOW THAT DICTIONARY IS PRCOESS OF STORING DATA IN FORM OF KEY:VALUE PAIR
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
# ---------------------------------------------------------------------------------------------------------------------------------
    # LIST FOR STORING THE DATA AND SET FOR HAVING THE UNIQUE VALUES ONLY
    clean_data = []
    unique_users = set()
# ---------------------------------------------------------------------------------------------------------------------------------
    for user in data:
        
        # 1. CLEANING PROCESS OF THE RATING 
        
        # i. getting the raw rating to get the idea of the actuall rating of the user from the data
        raw_rating = str(user["rating"]).strip().lower()
        # ii. now checking if i am having a raw rating present in the json data --> then converting it to number format
        if (raw_rating in text_to_num):
            # REPLACE THE VALUE FROM STRING TO NUMBER
            raw_rating = text_to_num[raw_rating]
        # iii. OVERWRITING THE USER DATA TO THE ACTUAL NUMBER FORMAT DATA
        user["rating"] = raw_rating
# ---------------------------------------------------------------------------------------------------------------------------------
        # 2. HANDLING OF THE MISSING VALUES
        raw_age = user.get("age")
        # CHECKING IF AGE IS THERE OR NOT AND IF NOT PRESENT THEN PUT NONE OVER THERE
        if (raw_age == None):
            user["age"] = None
# ---------------------------------------------------------------------------------------------------------------------------------
        # 3. DE_DUPLICATION (REMOVAL OF THE DUPLICATE DATA HERE)👇
        if (user["name"].strip() in unique_users):
            # if in the set we have the A user present then not to add in the list , and skip it
            continue
        
        # ADDITON OF THE USER IN THE SET
        unique_users.add(user["name"])
        # ADDING OF THE NON-DUPLICATE DATA IN THE LIST
        clean_data.append(user)
        
    return(clean_data)
```

```python
# CALLING OF THE ACTIONS ON DATA FUNCTION AND PASSING THE DATA IN IT
cleaned_data = clean_data(data)
print(cleaned_data)
```

```python
# GETTING MEANINGFUL INSIDE OF THE DATA FUNCTION
def get_insights(data):
    # 1. AVG RATING
    total_rating = 0
    for user in data:
        total_rating = total_rating + float(user["rating"])
    avg_rating = total_rating/len(data)
    print(avg_rating)

    # 2. %AGE OF USER WITH POOR RATING
    poor_rating = 0
    for user in data:
        # PUTTING A CONDITION TO CHECK FOR IT
        if (float(user["rating"]) < 3):
            poor_rating += 1
    print("percentage of user with poor rating: = " , (poor_rating/len(data))*100,"%")
```

```python
get_insights(cleaned_data)
```

```python
# BRAND RECOMMENDATION FEATURE ::
def give_recommendation(data):
    recommendation = []
    
    for user in data:
        current_recomendation = {}
        current_recomendation["name"] = user["name"]
        
        if(float(user["rating"])>=4):
            current_recomendation["brand"] = "Apple"
        else:
            current_recomendation["brand"] = "Samsung"
        
        recommendation.append(current_recomendation)
    return recommendation
```

```python
give_recommendation(data)
```
