```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
# You start from page 1 of the website.
page_count = 1

# You create an infinite loop that will run until you manually break it.
while True:

    # Creates the webpage link by inserting the page number.
    URL = f"https://quotes.toscrape.com/page/{page_count}/"

    # Sends a GET request to the URL.
    response = requests.get(URL)

    # Converts the raw HTML into a BeautifulSoup object.
    soup = BeautifulSoup(response.text, "lxml")

    # Selects all elements with <div class="quote">.
    quotes = soup.select("div.quote")

    # If quotes is empty → it means page does not exist.
    if not quotes:
        print("no valid pages anymore..")
        break

    # Opens (or creates) a file like:
    # SCRAPPED_DATA_FOLDER/quotes1.html,
    # SCRAPPED_DATA_FOLDER/quotes2.html, etc.
    with open(f"SCRAPPED_DATA_FOLDER/quotes{page_count}.html", "w") as f:
        # If you write: quotes1.txt → Python creates a TXT file
        # If you write: quotes1.json → Python creates a JSON file
        # If you write: quotes1.html → Python creates an HTML file
        # If you write: quotes1.hello → Python creates a file with extension .hello 
        # Python sees the filename ending with .html.
        # You opened it in "w" mode → means create this file.
        # So Python creates a new file with that extension.
       
        
        # Writes the entire HTML page into that file.
        f.write(response.text)
        
        # Prints a confirmation that the page was saved.
        print(f"downloaded data from page{page_count}")

    page_count += 1
```

```python
# EXTRACTION OF USEFULL INFORMATIONS FROM THE DATA STORED ::: 👇
with open("SCRAPPED_DATA_FOLDER/quotes1.html") as f:
    html_content = f.read()
    
soup = BeautifulSoup(html_content , "lxml")
```

```python
all_quotes = soup.select("div.quote")
life_quotes = []

for q in all_quotes:
    all_tags = []
    
    #CHECKING IN THE TAG LIKE THIS 
    for tag in q.select(".tags .tag"):
        all_tags.append(tag.get_text())
        
    if "live" in all_tags:
        text = q.select_one("span.text").get_text()
        author = q.select_one("small.author").get_text()
        life_quotes.append([author , text])
        
print(life_quotes)
```
