# SEABORN 🌊


```python
# IMPORTING SEABORN & MATPLOTLIB TOGETHER LIKE THIS
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# TO WORK WITH THE SEABORN ,

# 1st WE HAVE TO SELECT A THEME

sns.set_theme()  # if we leave this blank then it will be set to default
```

```python
# TO GET THE NAMES OF ALL THE DATASETS THEN WE WILL DO IT LIKE THIS
sns.get_dataset_names()
```

```python
# 2nd AFTER SETTING THE THEME THEN WE WILL BE PLOTTING THE DATASET NOW.

# LOADING OF THE TIPS DATASET IN THE VARIABLE LIKE THIS
tips = sns.load_dataset("tips")

# PRINTING THE TYPE OF THE DATASET THAT WE ARE LOADING
print(type(tips))

# PRINTING THE HEAD TO UNDERSTAND , ON WHICH DATASET I AM DEALING WITH
print(tips.head())
```

```python
# NOW FOR PLOTTING DO IT LIKE THIS ::

# TO WORK WITH THE THEME THEN WE WILL DO IT LIKE THIS ::
# sns.set_theme(style="dark")

# WE WILL BE PASSING SOME OF THE PARAMETERS IN THE RELPLOT LIKE THIS
sns.relplot(
    # WE HAVE TO TELL THE SEABORN 1st LIKE WHICH DATA I WILL BE PLOTTING HERE
    data=tips,

    # x = tips[total_bill] -> we dont have to write like this bcz , this is SEABORN
    # when written directly we will be getting the total_bill from the tips dataset automatically by the seaborn itself

    # seaborn will automatically figure out that , first select tips dataset ->
    # then form that select the total_bill columns

    x="total_bill",  # INDEPENDENT VARIABLE
    y="tip",  # DEPENDENT ON INDEPENDENT VARIABLE

    # TO HIGHLIGHT A PARTICULAR DATASETS THEN WE WILL DO IT LIKE THIS
    hue="smoker",

    # TO HANDLE THE SIZE OF THE POINTS OF OUR DATASETS THEN WE WILL DO IT LIKE THIS
    size="size",

    # IF U WANT DIFFERNT STYLE FOR A PARTICULAR COLUMN ::
    # style = "name of the column"
    style="smoker"


)
```

```python
# NOW FOR PLOTTING DO IT LIKE THIS ::

# TO WORK WITH THE THEME THEN WE WILL DO IT LIKE THIS ::
# sns.set_theme(style="dark")

# WE WILL BE PASSING SOME OF THE PARAMETERS IN THE RELPLOT LIKE THIS
sns.relplot(
    # WE HAVE TO TELL THE SEABORN 1st LIKE WHICH DATA I WILL BE PLOTTING HERE
    data=tips,

    # x = tips[total_bill] -> we dont have to write like this bcz , this is SEABORN
    # when written directly we will be getting the total_bill from the tips dataset automatically by the seaborn itself

    # seaborn will automatically figure out that , first select tips dataset ->
    # then form that select the total_bill columns

    x="total_bill",  # INDEPENDENT VARIABLE
    y="tip",  # DEPENDENT ON INDEPENDENT VARIABLE

    # TO HIGHLIGHT A PARTICULAR DATASETS THEN WE WILL DO IT LIKE THIS
    hue="smoker",

    # TO HANDLE THE SIZE OF THE POINTS OF OUR DATASETS THEN WE WILL DO IT LIKE THIS
    size="size",

    # IF U WANT DIFFERNT STYLE FOR A PARTICULAR COLUMN ::
    # style = "name of the column"
    style="smoker",

    # IF U WANT TO CHANGE THE PLOT STYLE THEN DO IT LIKE THIS
    kind="line"

)
```

MULTIPLE PLOTS WITH THE SEABORN


```python
# 1. SCATTER PLOTS ::
tips = sns.load_dataset("tips")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
```

```python
# 2 LINE PLOTS ::

tips = sns.load_dataset("tips")

sns.lineplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
```

CATEGORICAL DATA


```python
# OVER THE DAYS HOW MUCH TIPS THEY GOT FROM THE MALE AND FEMALE CUSTOMERS.

# FOR BARPLOT , IT WILL BE DONE LIKE THIS
sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
# FOR BOX PLOT WE WILL DO IT LIKE THIS ::
sns.boxplot(
    data=tips,
    x="day",
    y="tip"
)
```

```python
# DISTRIBUTION PLOTS :: HISTOGRAMS
penguins = sns.load_dataset("penguins")

# TO CREATE HISTOGRAMS WE DO IT LIKE THIS ::
sns.histplot(
    data=penguins,
    x="body_mass_g",
    bins=30
)
```

BEST WAY OF PLOTTING THE PLOTS USING THE SEABORN :: 

```python
# USING THE SAME THING OF THE MODERN MATPLOTLIB

fig, ax = plt.subplots()  # WOKRING LIKE MODERN MATPLOTLIB

sns.lineplot(  # WORKING LIKE THE SEABORN
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    errorbar = None,
    ax=ax  # AXIS IS SET SAME AS THE "AXIS"
)

ax.set_title("avg bill value for each day")
ax.set_xlabel("DAYS")
ax.set_ylabel("Total Bill Amount")

fig.tight_layout()
```
