```python
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25],  # New York
    [18, 19, 17, 20, 21],  # London
    [30, 32, 31, 33, 34],  # Delhi
    [25, 26, 24, 27, 28]   # Tokyo
]
fig, ax = plt.subplots(2, 2, figsize=(5, 5))

# our combinations will be somthing like this
# (0 , 0) & (0 , 1) & (1 , 0) & (1 , 1)

# city_no = 0
# for i in range(2):  # rows(i) - 0 , 1
#     for j in range(2):  # col(j) - 0 , 1
#         ax[i][j].plot(days, temperatures[city_no], marker='o')
#         ax[i][j].set_title(cities[city_no])
#         ax[i][j].grid(True)
#         city_no += 1

# 💠 IN PLACE OF THE ABOVE COMPLEX LOOP OF 2-D ARRAY I AM
# MAKINK THE ARRAY TO 1-D SHAPE AND WORKING ON IT LIKE THIS
ax = ax.flatten()
for i, ax in enumerate(ax):
    ax.plot(days, temperatures[i], marker='o')
    ax.set_title(cities[i])
    ax.grid(True)


fig.suptitle("TEMPERATURE DATA FOR IMPORTANT CITIES OF THE WORLD")
fig.tight_layout()
```
