```python
import psutil

ram = psutil.virtual_memory()
total = ram.total / (1024**3)
used = ram.used / (1024**3)
available = ram.available / (1024**3)

print(f"Total RAM given by Colab: {total:.2f} GB")
print(f"Used RAM right now: {used:.2f} GB")
print(f"Available RAM right now: {available:.2f} GB")
```
