```python
# FIRST U HAVE TO IMPORT THE PANDAS LIKE THIS 
import pandas as pd
```

```python
# THIS IS HOW A SERIES IS MADE
s = pd.Series([1, 2, 3, 4, 5])
print(s)

# THIS IS HOW U CAN ACCESS THE INDIVIDUAL VALUES IN THE SERIES USEING THEIR INDEX , like this
print(s[0])
print(s[3])
```

```python
# THIS IS HOW U CAN DO OR HAVE YOUR PERSONALISED LABELS
age = pd.Series([21, 22, 23, 24, 25] ,index = ["parth","vaibhav","parthiv","urjit","sarthak"])

# print(age["parth"])
# print(age["urjit"])
# print(age.index)

# THIS IS HOW U CAN DROP THE VALUES IN THE SERIES 
print("THIS BELOW IS THE ORIGINAL ONE BEFORE THE DROP 👇\n", age)
changed_age = age.drop("parth")

print("THIS IS AFTER THE DROP 👇 \n",changed_age)
```

```python
# DATA_FRAME

# 1. USING DICTIONARY :- 👇
info = {
    "name": ["Parth", "vaibhav", "urjit", "sarthak"],
    "age": [21, 20, 10, 11],
    "marks": ["100%", "10%", "90%", "1%"]
}
df = pd.DataFrame(info)
print(df)
df

# 2. LIST OF LIST :- 👇
df = pd.DataFrame([["Parth", 21], ["Parthiv", 20], [
                  "Urjit", 11]], columns=["Name", "Age"])
df

# 3. NUMPY ARRAYS :- 👇
import numpy as np
np_arr = np.array([[1, 2, 3], [2, 4, 6]])
df = pd.DataFrame(np_arr, columns=["A", "B", "C"])
df
```
