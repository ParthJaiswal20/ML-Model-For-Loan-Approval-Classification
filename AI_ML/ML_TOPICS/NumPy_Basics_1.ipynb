```python
# 1. TO HAVE THE NUMPY ARRAY , DECLARE THE SIMPLE LIST LIKE THIS [1,2,3,4,5] and THEN ADD THE NAME OF THE NUMPY (np) WITH
# . DOT OPERATOR AND ARRAY NAME , LIKE THIS
arr1 = np.array([1, 2, 3, 4, 5])
print(arr1)
print(type(arr1))
# 2. NOW ADDING A STRING IN THE NUMPY ARRAY WE HAVE THE WHOLE ARRAY AS THE STRING , like this
arr2 = np.array([1, 2, 3, 4, "parth"])
print(arr2)
print(type(arr2))

# 3. CREATION OF THE 2-D ARRAY
arr_2d = np.array([[1, 2], [3, 4], [5, 6]])
print(arr_2d, arr_2d.shape)
```
