```python
# GROUPING & AGGREGATION OF DATA ::
# 1. FIRST U HAVE TO GROUP THE DATA 
# df.groupby("country")

# 2. THEN U HAVE TO USE THE AGGREGATION FUNCTION WITH THE GROUPBY AS LIKE IT IS A CONDITON 
# ON WHAT BASIS U HAVE TO GROPU THEM , LIKE THIS 
# df.groupby("country")["income"].mean()

# U CAN USE MORE AGGREGATINON FUNCTIONS LIKE THIS 
# 1. SUM 
# df.groupby("country")["income"].sum()
# 2. MAX
# df.groupby("country")["income"].max()
# 3. MIN 
df.groupby("country")["income"].min()
# 4. COUNT 
df.groupby("country")["income"].count()

# ON THE BASIS OF GENDER IF I WANT TO FIND THE MEAN THEN DO IT LIKE THIS ::
df.groupby("gender")["income"].mean()

# FINDING THE SUM OF THE EACH GENDER INCOME 
df.groupby("gender")["income"].sum()

# FINDING THE MEAN OF THE AGE OF EACH GENDER LIKE THIS 
df.groupby("gender")["age"].mean()

# FINDING THE MEAN OF THE AGE OF EACH COUNTRY LIKE THIS 
df.groupby("country")["age"].mean()

# USE OF AGG IS LIKE THIS ::
df.groupby("country")["income"].agg(mean_salary = "mean" , min_salary = "min" , max_salary = "max" , SUM = "sum" , COUNT = "count")

# PERFORMING DIFF DIFF AGGREGATION FUNCTIONS ON DIFF DIFF COLUMNS TOGETHER LIKE THIS 
df.groupby("country").agg({
    "income" : "mean",
    "age" : "max"
})
```
