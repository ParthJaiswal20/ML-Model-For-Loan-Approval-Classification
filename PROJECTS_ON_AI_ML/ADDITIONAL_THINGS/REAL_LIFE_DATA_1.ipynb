```python
import pandas as pd
```

```python
# WHILE TRYING TO LOAD ANY DATA IN THE CODE , THE DATA AND THE PLACE OF CODING MUST BE SAME
# LOADING OF THE AIR QUALITY DATA IN THE DATA_FRAME
df = pd.read_csv("globalAirQuality.csv")
# PRINTING OF THE DF
df
```

```python
# SELECTING DATA 👇
df["city"]
df[["city", "aqi"]]

# ACCESSING THE ROW AND PRINTING ITS DATA
df.loc[0]
# PERFORMING THE SLICING OPERATION ON THE ROW USING THE LOC LIKE THIS
df.loc[2:4]

# SLICING OPERATION IN THE ROW USING THE ILOC
df.iloc[2:4]

# ACCESSING THE INDIVIDUAL ROW & COLUMN TOGETHER , LIKE THIS
df.loc[0,"aqi"]
# FOR MORE DETAILED THEN DO IT LIKE THIS
df.loc[0,["city","aqi"]]

# OR IF U WANT U CAN PRINT THIS ABOVE CONDITON TO UPTO SOME SPECIFIC ROWS , LIKE THIS
df.loc[0:2 , ["city","aqi"]]

# USING WITH ILOC 👇
# BOTH THE ROW & COL MUST BE IN INTEGER FORMAT ONLY
df.iloc[0:2 , 2]

# USING I_LOC VS LOC
# 1. LOC
df.loc[0:2, ["country", "city", "longitude", "latitude"]]
# 2. I_LOC
df.iloc[0:3, 1:5 ]

# ACCESSING A SPECIFIC CELL using IAT & AT
# 1. at
df.at[0, "city"]

# 2. i_at
df.iat[0 , 2]
```

```python
# FILTERING OF THE DATA :-

# METHOD - 1 :: 👇
# 1. SINGLE CONDITION
df[df["aqi"] > 100]

# 2. MULTIPLE CONDITION ALL TOGETHER
df[(df["aqi"] > 100) & (df["temperature"] > 30)]

# 3. MORE SPECIFIC WAY OF GETTING THE DESIRED DATA THAT I WANT
df[(df["aqi"] > 100) & (df["temperature"] > 30)][["city", "aqi"]]
```

```python
# QUERY METHOD :: 👇
df.query("aqi > 100")

# MULTIPLE QUERIES CONDITIONS LIKE THIS 👇
df.query("aqi > 100 & temperature > 30 ")

# CHAINING CAN ALSO BE DONE LIKE THIS 👇
df.query("aqi > 100 & temperature > 30 ")[["city", "aqi"]]

# USING PYTHON VARIABLE FOR THE REFERENCE WITH THE @ SYMBOL 👇
aqi_value_to_check = 100
df.query("aqi > @aqi_value_to_check & temperature > 30 ")[["city", "aqi"]]
```
